```python
class Person:
    def __init__(self, name, parents=()):
        self.name = name
        self.parents = parents # 这里面存的是Person对象，而不是名字
```

```python
granpa = Person('爷爷')
dad = Person('爸爸',(granpa,)) # 这里只有一个元素的tupel必须带逗号，不然python会把括号当成普通的数学符号。
me = Person('我',(dad,))
```

```python
print(me.parents[0].name)
```

```python
print(me.parents[0].parents[0].name)
```

# python的闭包形式
闭包这个形式，是用来解决一个时间的错位问题。
1、函数可以当成value传递，函数里面可以再定义函数。
2、函数里面可以再定义函数，内层函数可以调用外层的变量

```python
def make_counter():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc          # 把内层函数交出去

c = make_counter()      # make_counter 已经执行完了
c()                     # 1                    # 2
```

实际解决的问题：把一段以后才执行的代码和执行时需要的数据打包在一起，不用转门写个类。

```python
def make_multiplier(n):
    def times(x):
        return x * n       # n 不是 times 自己的参数，是外面的
    return times
```

```python
double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))
```

因为double返回的是一个函数，也就是time(x),所以double(5)的时候，就是times(5)

```python
print(triple(5))
```

事情的关键是，make_multiplier这个函数执行完了，但是n还在

```python
def make_piggy_bank():
    total = 0
    def save(money):
        nonlocal total     # 声明：我要改的是外面那个 total
        total += money
        return total
    return save

xiaoming = make_piggy_bank()
xiaohong = make_piggy_bank()
```

```python
print(xiaoming(10))   # 10
print(xiaoming(5))    # 15   ← 记得上次的 10
print(xiaohong(100))  # 100  ← 小红有自己的 total，和小明互不影响
print(xiaoming(1))    # 16
```
